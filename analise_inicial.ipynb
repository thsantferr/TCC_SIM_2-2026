## Carregar base completa

```python
from pathlib import Path
import re

# Para anos que têm DOSP* (estadual SP), preferir esse arquivo e descartar
# Mortalidade_Geral_* e DO**OPEN (bases nacionais/preliminares) do mesmo ano.
# Isso evita duplicatas após o filtro por CODMUNRES.

def _extrair_ano(nome):
    m4 = re.search(r'(\d{4})', nome)
    if m4:
        return int(m4.group(1))
    m2 = re.match(r'DO(\d{2})OPEN', nome)
    if m2:
        return 2000 + int(m2.group(1))
    return None

dosp_anos = {
    _extrair_ano(Path(p).stem)
    for p in dt
    if Path(p).stem.startswith('DOSP')
}

arquivos_carregar = []
for p in dt:
    nome = Path(p).stem
    ano  = _extrair_ano(nome)
    if (nome.startswith('Mortalidade_Geral_') or re.match(r'DO\d{2}OPEN', nome)) and ano in dosp_anos:
        print(f"Pulando {nome} — DOSP{ano} disponível")
        continue
    arquivos_carregar.append(p)

print(f"\nArquivos selecionados: {len(arquivos_carregar)} de {len(dt)}")
print(f"Anos com DOSP (estadual SP): {sorted(dosp_anos)}")
```

Filtro de CODMUNRES incorreto (causa principal da ausência de 2000–2005)

  Os arquivos SIM de 2000 a 2005 armazenam CODMUNRES com 7 dígitos (ex. 3550308 para São Paulo), incluindo o dígito verificador do IBGE. A partir de 2006, o DATASUS passou a usar 6 dígitos (ex. 355030).   
                                                                                                                                                                                                  
  O filtro atual no notebook (>= 350000 e < 360000) só captura a versão de 6 dígitos — zerando completamente os anos 2000–2005 (~1,43 milhão de óbitos perdidos)

```python
lst = []
for path in arquivos_carregar:
    df = pd.read_parquet(path, engine='pyarrow')
    df.columns = [c.upper().strip() for c in df.columns]
    if 'CODMUNRES' in df.columns:
        df['CODMUNRES'] = df['CODMUNRES'].str.slice(0,6) 
        df['CODMUNRES'] = pd.to_numeric(df['CODMUNRES'])
        df = df[(df['CODMUNRES'] >= 350000) & (df['CODMUNRES'] < 360000)]
    lst.append(df.copy())
    del df
df = pd.concat(lst, axis=0, ignore_index=True)
del lst
```

# Visualizações — Análise Epidemiológica SP (2000–2025)

```python
import duckdb
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

PARQUET = 'variavel/dataframe_tratado.parquet'
con = duckdb.connect()

plt.rcParams.update({
    'font.family': 'DejaVu Sans',
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.grid': True,
    'axes.axisbelow': True,
    'grid.color': '#e5e7eb',
    'grid.linewidth': 0.8,
    'figure.dpi': 110,
    'figure.facecolor': 'white',
    'axes.facecolor': 'white',
})
print("Setup OK.")
```

```python
# ── Figura 1: Série temporal de óbitos totais ────────────────────────────────
df_ano = con.execute(f"""
    SELECT year(DTOBITO) AS ano, COUNT(*) AS obitos
    FROM read_parquet('{PARQUET}')
    WHERE year(DTOBITO) BETWEEN 2000 AND 2025
    GROUP BY 1 ORDER BY 1
""").df()

fig, ax = plt.subplots(figsize=(12, 5))
COR = '#1d6fa5'
ax.plot(df_ano['ano'], df_ano['obitos'], color=COR, linewidth=2.5,
        marker='o', markersize=5, zorder=3, clip_on=False)
ax.fill_between(df_ano['ano'], df_ano['obitos'], alpha=0.10, color=COR)

# Bandas de eventos sanitários
ax.axvspan(2009, 2009.9, alpha=0.13, color='#e07b39', zorder=1)
ax.text(2009.05, df_ano['obitos'].max() * 0.985, 'H1N1', fontsize=8,
        color='#b34a00', va='top', fontweight='bold')
ax.axvspan(2019.9, 2022.1, alpha=0.13, color='#c0392b', zorder=1)
ax.text(2020.05, df_ano['obitos'].max() * 0.985, 'COVID-19', fontsize=8,
        color='#7b241c', va='top', fontweight='bold')

# Anotação do pico
pico = df_ano.loc[df_ano['obitos'].idxmax()]
ax.annotate(f"{int(pico['obitos']):,} óbitos",
            xy=(pico['ano'], pico['obitos']),
            xytext=(pico['ano'] - 3, pico['obitos'] * 0.965),
            arrowprops=dict(arrowstyle='->', color='#7b241c', lw=1.2),
            fontsize=8.5, color='#7b241c')

# Indicador 2025 incompleto
ax.axvline(x=2024.92, color='#9ca3af', linestyle='--', linewidth=1)
ax.text(2024.95, df_ano['obitos'].min() * 1.01, '2025*', fontsize=7.5,
        color='#9ca3af', va='bottom')

ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x/1000:.0f}k'))
ax.set_xlim(1999.5, 2025.8)
ax.set_xlabel('Ano', labelpad=6)
ax.set_ylabel('Óbitos', labelpad=6)
ax.set_title('Óbitos totais por ano — Estado de São Paulo (2000–2025)',
             fontsize=12, fontweight='bold', loc='left', pad=10)
ax.text(0, -0.13, '* 2025 possivelmente incompleto (viés de registro tardio).',
        transform=ax.transAxes, fontsize=7.5, color='#6b7280')
plt.tight_layout()
plt.savefig('variavel/fig_serie_temporal.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── Figura 2: Composição proporcional por grupo CID ──────────────────────────
CASO = """
    CASE
        WHEN CAUSABAS LIKE 'A%' OR CAUSABAS LIKE 'B%'             THEN 'Infecciosas'
        WHEN CAUSABAS LIKE 'C%'                                    THEN 'Neoplasias'
        WHEN CAUSABAS LIKE 'I%'                                    THEN 'Cardiovasculares'
        WHEN CAUSABAS LIKE 'J%'                                    THEN 'Respiratórias'
        WHEN CAUSABAS LIKE 'V%' OR CAUSABAS LIKE 'W%'
          OR CAUSABAS LIKE 'X%' OR CAUSABAS LIKE 'Y%'             THEN 'Causas externas'
        WHEN CAUSABAS LIKE 'R%'                                    THEN 'Mal definidas'
        ELSE                                                            'Demais crônicas'
    END
"""

df_g = con.execute(f"""
    SELECT year(DTOBITO) AS ano, {CASO} AS grupo, COUNT(*) AS obitos
    FROM read_parquet('{PARQUET}')
    WHERE year(DTOBITO) BETWEEN 2000 AND 2025
    GROUP BY 1, 2 ORDER BY 1, 2
""").df()

df_pct = (df_g.pivot(index='ano', columns='grupo', values='obitos')
              .fillna(0)
              .apply(lambda r: r / r.sum() * 100, axis=1))

CORES = {
    'Cardiovasculares': '#1d6fa5',
    'Neoplasias':       '#e07b39',
    'Respiratórias':    '#2d9f6a',
    'Infecciosas':      '#9b59b6',
    'Causas externas':  '#c0392b',
    'Mal definidas':    '#7f8c8d',
    'Demais crônicas':  '#d4a017',
}
ordem = [g for g in CORES if g in df_pct.columns]

fig, ax = plt.subplots(figsize=(13, 6))
bottom = np.zeros(len(df_pct))
anos = df_pct.index.values

for g in ordem:
    vals = df_pct[g].values
    ax.bar(anos, vals, bottom=bottom, color=CORES[g], label=g, width=0.8, linewidth=0)
    bottom += vals

ax.axvspan(2019.6, 2022.4, alpha=0.12, color='#111', zorder=0)
ax.text(2019.65, 98, 'COVID-19', fontsize=7.5, color='#333', va='top')

ax.set_ylim(0, 100)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x:.0f}%'))
ax.set_xlabel('Ano', labelpad=6)
ax.set_ylabel('Proporção (%)', labelpad=6)
ax.set_title('Composição proporcional dos óbitos por grupo de causa — SP (2000–2025)',
             fontsize=12, fontweight='bold', loc='left', pad=10)
ax.legend(loc='lower left', frameon=False, fontsize=8.5, ncol=2,
          bbox_to_anchor=(0.0, -0.25))
plt.tight_layout()
plt.savefig('variavel/fig_grupos_proporcao.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── Figura 3: Excesso de mortalidade mensal 2020–2022 ────────────────────────
df_mes = con.execute(f"""
    SELECT year(DTOBITO) AS ano, month(DTOBITO) AS mes, COUNT(*) AS obitos
    FROM read_parquet('{PARQUET}')
    WHERE year(DTOBITO) BETWEEN 2015 AND 2022
    GROUP BY 1, 2 ORDER BY 1, 2
""").df()

baseline = (df_mes[df_mes['ano'].between(2015, 2019)]
            .groupby('mes')['obitos'].mean())

df_cv = df_mes[df_mes['ano'].between(2020, 2022)].copy().reset_index(drop=True)
df_cv['esperado'] = df_cv['mes'].map(baseline)
df_cv['excesso']  = df_cv['obitos'] - df_cv['esperado']
df_cv['label']    = df_cv['ano'].astype(str) + '-' + df_cv['mes'].astype(str).str.zfill(2)

fig, ax = plt.subplots(figsize=(13, 5))
cores = ['#c0392b' if v >= 0 else '#1d6fa5' for v in df_cv['excesso']]
ax.bar(df_cv.index, df_cv['excesso'], color=cores, width=0.72, linewidth=0)
ax.axhline(0, color='#374151', linewidth=1)

# Divisores de ano
for ano in [2021, 2022]:
    idx = df_cv.index[df_cv['ano'] == ano]
    if len(idx):
        ax.axvline(idx[0] - 0.5, color='#d1d5db', linewidth=0.8)

# Rótulos de ano centralizados
for ano in [2020, 2021, 2022]:
    idxs = df_cv.index[df_cv['ano'] == ano]
    if len(idxs):
        mid = (idxs[0] + idxs[-1]) / 2
        ax.text(mid, ax.get_ylim()[0] * 0.88, str(ano),
                fontsize=9, color='#6b7280', ha='center', fontweight='bold')

total = int(df_cv['excesso'].sum())
ax.text(0.99, 0.97, f'Excesso total 2020–2022: {total:+,.0f} óbitos',
        transform=ax.transAxes, ha='right', va='top',
        fontsize=9, color='#7b241c', fontweight='bold')

step = max(1, len(df_cv) // 12)
ax.set_xticks(df_cv.index[::step])
ax.set_xticklabels(df_cv['label'].iloc[::step], rotation=45, ha='right', fontsize=7.5)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x:+,.0f}'))
ax.set_xlabel('Mês', labelpad=6)
ax.set_ylabel('Δ óbitos vs. baseline', labelpad=6)
ax.set_title('Excesso de mortalidade mensal — SP (2020–2022)',
             fontsize=12, fontweight='bold', loc='left', pad=10)
ax.text(0, -0.18, 'Baseline = média mensal 2015–2019. Vermelho = acima do esperado, azul = abaixo.',
        transform=ax.transAxes, fontsize=7.5, color='#6b7280')
plt.tight_layout()
plt.savefig('variavel/fig_excesso_mortalidade.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── Figura 4: Envelhecimento dos óbitos e composição por sexo ────────────────
df_dem = con.execute(f"""
    SELECT
        year(DTOBITO) AS ano,
        PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY IDADE) AS idade_mediana,
        SUM(CASE WHEN SEXO = '1' THEN 1 ELSE 0 END) * 100.0 / COUNT(*) AS pct_masc,
        SUM(CASE WHEN SEXO = '2' THEN 1 ELSE 0 END) * 100.0 / COUNT(*) AS pct_fem
    FROM read_parquet('{PARQUET}')
    WHERE year(DTOBITO) BETWEEN 2000 AND 2025
      AND IDADE IS NOT NULL AND IDADE BETWEEN 0 AND 120
      AND SEXO IN ('1', '2')
    GROUP BY 1 ORDER BY 1
""").df()

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 7), sharex=True,
                                gridspec_kw={'hspace': 0.08})

COR_I = '#1d6fa5'
COR_S = '#e07b39'
COR_F = '#FF8C69'

# Painel 1 — Idade mediana
ax1.plot(df_dem['ano'], df_dem['idade_mediana'], color=COR_I, linewidth=2.5,
         marker='o', markersize=4)
ax1.fill_between(df_dem['ano'], df_dem['idade_mediana'],
                 df_dem['idade_mediana'].min() - 0.5, alpha=0.10, color=COR_I)
delta = df_dem['idade_mediana'].iloc[-1] - df_dem['idade_mediana'].iloc[0]
ax1.text(0.01, 0.93, f'▲ {delta:.1f} anos no período (envelhecimento)',
         transform=ax1.transAxes, fontsize=8.5, color=COR_I, va='top')
ax1.set_ylabel('Idade mediana ao óbito (anos)', labelpad=6)
ax1.set_title('Envelhecimento dos óbitos e proporção por sexo — SP (2000–2025)',
              fontsize=12, fontweight='bold', loc='left', pad=8)

# Painel 2 — % masculino
ax2.plot(df_dem['ano'], df_dem['pct_masc'], color=COR_S, linewidth=2.5,
         marker='o', markersize=4,label='Homens')
ax2.plot(df_dem['ano'], df_dem['pct_fem'], color=COR_F, linewidth=2.5,
         marker='o', markersize=4,label='Mulheres')
ax2.axhline(50, color='#9ca3af', linestyle='--', linewidth=1, alpha=0.8)
ax2.text(2000.3, 50.4, '50%', fontsize=7.5, color='#9ca3af')
ax2.set_ylabel('% óbitos por sexo', labelpad=6)
ax2.set_xlabel('Ano', labelpad=6)
ax2.legend()

# Bandas de eventos nos dois painéis
for ax in (ax1, ax2):
    ax.axvspan(2009, 2009.9,  alpha=0.10, color='#e07b39', zorder=0)
    ax.axvspan(2019.9, 2022.1, alpha=0.10, color='#c0392b', zorder=0)

ax2.text(2020.1, df_dem['pct_masc'].max() * 0.998,
         'COVID-19', fontsize=7.5, color='#7b241c', va='top')

plt.tight_layout()
plt.savefig('variavel/fig_demografico.png', dpi=150, bbox_inches='tight')
plt.show()
```
