# Coleta de dados

```python
anos_coleta = list(range(2000,2026))
print(anos_coleta)
```

## coleta de bases

```python
#download dados de 2015 a 2025
dt = sim(state='SP',year=anos_coleta)#,as_dataframe=True)
```

```python
del df1, df2, df3, val
```

## carregar base completa

```python
lst = []
for i in dt:
    df = pd.read_parquet(i,engine='pyarrow')
    df.columns = [i.upper().strip() for i in df.columns]
    if"CODMUNRES" in df.columns:
        df["CODMUNRES"] = pd.to_numeric(df["CODMUNRES"])
        df = df[(df["CODMUNRES"]>=350000)&(df["CODMUNRES"]<360000)]
    lst.append(df.copy())
    del df
df = pd.concat(lst,axis=0,ignore_index=True)
del lst
```

```python
df.shape
```

```python
df.sample(15).transpose().to_csv('./verificacao_exemplos.csv',sep=';',decimal=',',encoding='utf8')
```
