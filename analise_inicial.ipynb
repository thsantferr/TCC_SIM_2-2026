## Carregar base completa

```python
lst = []
for i in dt:
    df = pd.read_parquet(i,engine='pyarrow')
    df.columns = [i.upper().strip() for i in df.columns]
    if"CODMUNRES" in df.columns:
        df["CODMUNRES"] = pd.to_numeric(df["CODMUNRES"])
        df = df[(df["CODMUNRES"]>=350000)&(df["CODMUNRES"]<360000)]
    lst.append(df.copy())
    del df
df = pd.concat(lst,axis=0,ignore_index=True)
del lst
```

## Selecionar colunas relevantes

Remove campos de baixo valor para o problema de pesquisa (transição epidemiológica e padrões de
mortalidade em SP): metadados de processamento/lote/cartório, campos de oportunidade de recebimento
do registro, naturalidade, linhas brutas do atestado (já resumidas em `CAUSABAS`), e os blocos
materno/fetal e de investigação de óbito (fora do escopo definido para este TCC — não haverá
subseção dedicada a mortalidade materna/infantil). Ver `dicionario_campos.md` para a descrição de
cada campo.

```python
colunas_remover = [
    # identificador unico não nescessario manter
    'CONTADOR',
    # Sistema, processamento, lote e cartório (não documentados / uso interno)
    'EXPDIFDATA', 'TP_ALTERA', 'CB_ALT', 'CODMUNCART', 'CODCART', 'NUMREGCART',
    'DTREGCART', 'CRM', 'TPASSINA', 'NUMEROLOTE', 'STCODIFICA', 'CODIFICADO',
    'VERSAOSIST', 'VERSAOSCB', 'DTRECEBIM', 'DTCADASTRO',

    # Oportunidade/recebimento do registro
    'DTRECORIG', 'DTRECORIGA', 'DIFDATA', 'OPOR_DO',

    # Naturalidade
    'NATURAL', 'CODMUNNATU',

    # Estabelecimento em texto (redundante com CODESTAB)
    'ESTABDESCR',

    # Causas — linhas brutas do atestado (já resumidas em CAUSABAS)
    'LINHAA', 'LINHAB', 'LINHAC', 'LINHAD', 'LINHAII', 'ATESTADO', 'CB_PRE', 'CAUSABAS_O',

    # Bloco materno/fetal — fora do escopo deste TCC
    'IDADEMAE', 'ESCMAE', 'ESCMAE2010', 'ESCMAEAGR1', 'SERIESCMAE', 'OCUPMAE',
    'QTDFILVIVO', 'QTDFILMORT', 'GRAVIDEZ', 'SEMAGESTAC', 'GESTACAO', 'PARTO',
    'OBITOPARTO', 'PESO', 'TPMORTEOCO', 'OBITOGRAV', 'OBITOPUERP', 'NUMERODN', 'CAUSAMAT',

    # Bloco de investigação de óbito materno/infantil — fora do escopo deste TCC
    'DTINVESTIG', 'FONTEINV', 'DTCADINV', 'DTCONINV', 'DTCADINF', 'NUDIASOBCO',
    'NUDIASOBIN', 'NUDIASINF', 'FONTES', 'FONTESINF', 'TPRESGINFO', 'TPNIVELINV',
    'TPPOS', 'DTCONCASO', 'ALTCAUSA', 'MORTEPARTO', 'TPOBITOCOR', 'MAT_CLAS',
]

df = df.drop(columns=colunas_remover)
df.shape
```
